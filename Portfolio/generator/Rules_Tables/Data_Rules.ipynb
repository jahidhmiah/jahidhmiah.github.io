```python
#Import packages
import pandas as pd 
import numpy as np
import datetime
import json
```

```python
#Create columns and data I want to hold for Lead Source rules. 

lead_source_data = [
        ['Sales', 0.05, 0.1], 
        ['Organic', 0.15, 0.25],
        ['Event', 0.1, 0.2],
        ['Content', 0.1, 0.2],
        ['Partnership', 0.25, 0.40],
        ['Referral', 0.25, 0.40],
        ]
```

```python
lead_source_columns = ['Name', 'Min', 'Max']
lead_columns = [
                'LeadID','Created Date', 'Lead Status', 'MQL Date', 
                'Qualifying Date', 'Nurture Date', 'Qualified Date',
                'Lead Source', 'Lead Source Details','Rep ID', 'Rep Name',
                'ReMQL Date', 'Recent Lead Source', 'Recent Lead Source Details'
                ]
opportunity_columns = [
                "OpportunityID", "AccountID", "LeadID", "RepID",
                "Rep Name", "Created Date", "Close Date", "Stage", 
                "Current Stage Age", "Prospect Date", "Qualification Date",
                "Demo Date", "Proposal Date", "Negotiation Date", 
                "OutForSignature Date", "Amount", "PipelineCategory"
                ]
account_columns = [
                "AccountID", "Industry", "Employee Count", "Account Name", 
                "BDR ID","BDR","AE ID", "AE"
                ]
campaign_columns = [
                "CampaignID", "Campaign Name", "Campaign Type", "Campaign Subtype", "Campaign Start Date"
                ]
campaign_member_columns = [
                "CampaignMemberID","CampaignID", "LeadID", "Date"
                ]
```

```python
lead_source_df = pd.DataFrame(lead_source_data, columns = lead_source_columns)
lead_df = pd.DataFrame(columns = lead_columns)
opportunity_df = pd.DataFrame(columns = opportunity_columns)
account_df = pd.DataFrame(columns = account_columns)
campaign_df = pd.DataFrame(columns = campaign_columns)
campaign_member_df = pd.DataFrame(columns = campaign_member_columns)
```
